```python
board=[''for _ in range(9)]
```

```python
def display_board():
  print()
  print(" " + (board[0] if board[0] else " ") + " | " +
        (board[1] if board[1] else " ") + " | " +
        (board[2] if board[2] else " "))

  print("---+---+---")

  print(" " + (board[3] if board[3] else " ") + " | " +
        (board[4] if board[4] else " ") + " | " +
        (board[5] if board[5] else " "))

  print("---+---+---")

  print(" " + (board[6] if board[6] else " ") + " | " +
        (board[7] if board[7] else " ") + " | " +
        (board[8] if board[8] else " "))
  print()
```

```python
def check_winner():
  winning_position=[
      (0,1,2),(3,4,5),(6,7,8),
      (0,3,6),(1,4,7),(2,5,8),
      (0,4,8),(2,4,6)
  ]

  for a,b,c in winning_position:
    if board[a]==board[b]==board[c] and board[a]!='':
      return board[a]

  if '' not in board:
    return 'Draw'

  return None
```

```python
def mimimax(is_minimizing):
  result = check_winner()

  if result == 'O':
    return 1
  elif result == 'X':
    return -1
  elif result == 'Draw':
    return 0

  if not is_minimizing:
    best_score = -float('inf')

    for i in range(9):
      if board[i] == '':
        board[i] = 'O'
        score = mimimax(True)
        board[i] = ''
        best_score = max(score, best_score)

    return best_score

  else:
    best_score = float('inf')

    for i in range(9):
      if board[i] == '':
        board[i] = 'X'
        score = mimimax(False)
        board[i] = ''
        best_score = min(score, best_score)

    return best_score
```

```python
def best_move():
  best_score=-999
  move=-1

  for i in range(9):
    if board[i]=='':
      board[i]='O'
      score=mimimax(False)
      board[i]=''

      if score>best_score:
        best_score=score
        move=i

  return move
```

```python
print("TIC-TAC_TOE")
print("You are X")
print("Computer is O")

while True:
  display_board()

  position=int(input("Enter position (1-9):"))-1

  if position<0 or position>8 or board[position]!='':
    print("Invalid move! Try again.")
    continue

  board[position]='X'
  result = check_winner()

  if result:
    display_board()

    if result=='Draw':
      print("Game Draw!")
    else:
      print(result+" wins!")

    break


  computer_position = best_move()

  board[computer_position] = 'O'

  print("Computer played position:", computer_position + 1)

  result = check_winner()

  if result:
    display_board()

    if result == 'Draw':
      print("Game Draw!")
    else:
      print(result + " wins!")

    break
```

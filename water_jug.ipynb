```python
from collections import deque
```

```python
jug1_capacity=4
jug2_capacity=3
```

```python
initial_state=(0,0)
```

```python
visited =set()
```

```python
queue = deque([initial_state])
```

```python
print("State Space")
```

```python
while queue:
    state=queue.popleft()
    if state in visited:
        continue
    visited.add(state)

    x,y =state
    print(state)

    next_states=[
        (jug1_capacity,y),
        (x,jug2_capacity),
        (0,y),
        (x,0)
    ]
    transfer =min(x,jug2_capacity-y)
    next_states.append((x-transfer,y+transfer))

    transfer =min(y,jug1_capacity-x)
    next_states.append((x+transfer,y-transfer))

    for next_state in next_states:
        if next_state not in visited:
            queue.append(next_state)

```

```python
print("\nTotal Reachable States=",len(visited))
```

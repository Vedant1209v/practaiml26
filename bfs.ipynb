```python
graph={
    'A':['B','C'],
    'B':['D','E'],
    'C':['F'],
    'D':[],
    'E':['F'],
    'F':[]
}
```

```python
from collections import deque

def bfs(graph,start,goal):
  queue = deque([[start]])
  visited = set()

  while queue:
    path = queue.popleft()
    node = path[-1]

    if node == goal:
      return path

    if node not in visited:
      visited.add(node)

      for neighbor in graph[node]:
        new_path = list(path)
        new_path.append(neighbor)
        queue.append(new_path)
  return None
```

```python
start= 'A'
goal= 'F'
result=bfs(graph,start,goal)

if result:
  print("Path from :","->" .join(result))
else:
  print("No path found")
```

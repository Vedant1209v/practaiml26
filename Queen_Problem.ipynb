```python
class QueenChessBoard:
    def __init__(self, size):
        self.size = size
        self.columns = []

    def place_in_next_row(self, column):
        self.columns.append(column)

    def remove_in_current_row(self):
        return self.columns.pop()

    def display(self):
        for row in range(self.size):
            for column in range(self.size):
                if column == self.columns[row]:
                    print("Q", end=" ")
                else:
                    print(".", end=" ")
            print()

    def is_this_column_safe_in_next_row(self, column):
        row = len(self.columns)

        for queen_column in self.columns:
            if queen_column == column:
                return False

        for queen_row, queen_column in enumerate(self.columns):
            if abs(column - queen_column) == abs(row - queen_row):
                return False

        return True
```

```python
  def solve_queen(size):
    board = QueenChessBoard(size)
    number_of_solutions = 0

    current_row = 0

    column_to_try = 0

    while True:

      while column_to_try < size:
        if board.is_this_column_safe_in_next_row(column_to_try):
          board.place_in_next_row(column_to_try)
          current_row += 1
          column_to_try = 0
          break
        else:
          column_to_try += 1


      if current_row == size:
        board.display()
        print()
        number_of_solutions += 1

        prev_column = board.remove_in_current_row()
        current_row -= 1
        column_to_try = prev_column + 1

      elif column_to_try == size:
        try:
          prev_column = board.remove_in_current_row() #
          current_row -= 1
          column_to_try = prev_column + 1
        except IndexError:

          break


    print("Number of solution:",number_of_solutions)
    return number_of_solutions
```

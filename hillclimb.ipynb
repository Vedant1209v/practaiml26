```python
import math
```

```python
point=[(1,5),(6,4),(5,2),(2,1)]
```

```python
x,y=1.0,1.0
step=0.1
```

```python
def total_distance(x,y):
  return sum((x-px)**2+(y-py)**2 for px,py in point)

current = total_distance(x,y)
iteration=1

```

```python
while True:
  print(f"Iteration {iteration}:({x:.1f},{y:.1f}) Distance = {current:.2f}")

  neighbors=[
      (x+step,y),
      (x-step,y),
      (x,y+step),
      (x,y-step)
  ]

  best_x,best_y=x,y
  best_distance=current

  for nx,ny in neighbors:
    d=total_distance(nx,ny)
    if d<best_distance:
      best_x,best_y=nx,ny
      best_distance=d

  if best_distance>=current:
    break

  x,y=best_x,best_y
  current=best_distance
  iteration+=1
```

```python
print("\nOptimal Point:",(round(x,2),round(y,2)))
print("Minimum Distance:",round(current,2))
```

```python
from simpleai.search import SearchProblem,astar
```

```python
GOAL="HELLO WORLD"
```

```python
class HelloProblem(SearchProblem):
    def __init__(self, initial_state, goal_string):
        super().__init__(initial_state=initial_state)
        self.goal_string = goal_string

    def actions(self, state):
        if len(state) < len(self.goal_string):
          return list(" ABCDEFGHIJKLMNOPQRSTUVWXYZ")
        else:
          return[]
    def result(self, state, action):
        return state + action
    def is_goal(self, state):
        return state == self.goal_string

    def heuristic(self, state):
        wrong = sum(1
                    for i in range(len(state))
                    if i < len(self.goal_string) and state[i] != self.goal_string[i]
        )
        missing=len(self.goal_string)-len(state)
        return wrong + missing
```

```python
problem = HelloProblem(initial_state="", goal_string=GOAL)
result = astar(problem)

if result:
    print("Goal State:", result.state)
    print("\nPath:")
    for action, state in result.path():
        print(action, "->", state)
```

```python
def print_board(board):
    for row in board:
        print(row)
    print()


def find_blank(board):
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                return i, j


def generate_moves(board):
    x, y = find_blank(board)

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    successors = []

    for dx, dy in moves:
        nx = x + dx
        ny = y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            # Create a copy of the board
            new_board = [row[:] for row in board]

            # Swap blank with adjacent tile
            new_board[x][y], new_board[nx][ny] = (
                new_board[nx][ny],
                new_board[x][y]
            )

            successors.append(new_board)

    return successors


# Initial board
board = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

print("Initial state:")
print_board(board)

print("Possible next states:")

for state in generate_moves(board):
    print_board(state)
```
